```python
import torch
import math
import numpy as np
import sys
import os
#import omnilearned
sys.path.append(os.path.expanduser("~/Downloads/Jet_project"))
from petmodel import qspace
import matplotlib.pyplot as plt
import itertools
from itertools import combinations

# %%
qhead="/Users/shashin/Downloads/SARGE/diffusion_runs_N5/diffusion_runs_qspace_roundtrip_T"
phats_5 = []
ts_forward_5 = [5, 30, 80, 150, 250, 500, 1000, 2000, 3000, 4000, 5000]
for i in ts_forward_5:
    ckpt = torch.load(qhead+str(i)+".pt",weights_only=True)
    qhats = ckpt['Q_rev'] if isinstance(ckpt, dict) else ckpt
    ps_temp = torch.zeros(qhats.shape)
    for j in range(ps_temp.shape[0]):
        ps_temp[j]=qspace.qs_to_ps(qhats[j])
    phats_5.append(ps_temp) #shape (checkpoints, trajectories, samples, particles, 3-momentum)



qhead="/Users/shashin/Downloads/SARGE/diffusion_runs_N10/diffusion_runs_N10_qspace_roundtrip_T"
phats_10 = []
ts_forward_10 = [5, 75, 150, 200 ,500, 1000, 2000, 3000, 4000, 5000]
for i in ts_forward_10:
    ckpt = torch.load(qhead+str(i)+".pt",weights_only=True)
    qhats = ckpt['Q_rev'] if isinstance(ckpt, dict) else ckpt
    ps_temp = torch.zeros(qhats.shape)
    for j in range(ps_temp.shape[0]):
        ps_temp[j]=qspace.qs_to_ps(qhats[j])
    phats_10.append(ps_temp) #shape (checkpoints, trajectories, samples, particles, 3-momentum)


qhead="/Users/shashin/Downloads/SARGE/diffusion_runs_N20/diffusion_runs_N20_xi_1.4/diffusion_runs_N20_qspace_roundtrip_T"
phats_20 = []
#ts_forward_20 = [5, 50, 100, 200, 350, 500, 700, 1000, 2000, 3000, 4000, 5000]
ts_forward_20 = [5, 50, 100, 200, 350 ,500, 700, 1000, 2000, 3000, 4000, 5000]
for i in ts_forward_20:
    ckpt = torch.load(qhead+str(i)+".pt",weights_only=True)
    qhats = ckpt['Q_rev'] if isinstance(ckpt, dict) else ckpt
    ps_temp = torch.zeros(qhats.shape)
    for j in range(ps_temp.shape[0]):
        ps_temp[j]=qspace.qs_to_ps(qhats[j])
    phats_20.append(ps_temp) #shape (checkpoints, trajectories, samples, particles, 3-momentum)
#print(phats[0][0,0])

qhead="/Users/shashin/Downloads/SARGE/diffusion_runs_N20/diffusion_runs_N20_xi_1.4/diffusion_runs_N20_qspace_roundtrip_T"
phats_25 = []
ts_forward_25 = [5, 50, 100, 200, 350, 500, 700, 1000, 2000, 3000, 4000, 5000]
for i in ts_forward_25:
    ckpt = torch.load(qhead+str(i)+".pt",weights_only=True)
    qhats = ckpt['Q_rev'] if isinstance(ckpt, dict) else ckpt
    ps_temp = torch.zeros(qhats.shape)
    for j in range(ps_temp.shape[0]):
        ps_temp[j]=qspace.qs_to_ps(qhats[j])
    phats_25.append(ps_temp) #shape (checkpoints, trajectories, samples, particles, 3-momentum)
#print(phats[0][0,0])

pspace_dat_5 = torch.load("/Users/shashin/Downloads/SARGE/pspace_dat_N5.pt",weights_only=True)
pspace_dat_10 = torch.load("/Users/shashin/Downloads/SARGE/pspace_dat_N10.pt",weights_only=True)
pspace_dat_20 = torch.load("/Users/shashin/Downloads/SARGE/pspace_dat_N20.pt",weights_only=True)
#pspace_dat_25 = torch.load("/Users/shashin/Downloads/SARGE/pspace_dat_N25.pt",weights_only=True)

def compute_momentum_susceptibility(p0,phat,norm=True): #computes at fixed time: phats is a (Ntraj,Nsample,Nparticles,3) array
    Ntraj=phat.shape[0]
    Nsample=phat.shape[1]
    DeltaPs = p0[None,:Nsample,...]-phat
    pnorms=DeltaPs.norm(dim=-1)
    # per-trajectory validity: a trajectory-event-particle is bad if any NaN/Inf
    valid = torch.isfinite(pnorms)                   # (T, N, P) bool
    pnorms = torch.where(valid, pnorms, torch.zeros_like(pnorms))  # zero out bad entries

# NaN-aware trajectory mean: sum over valid, divide by valid count
    cnt = valid.sum(dim=0).clamp_min(1)              # (N, P) — how many good trajectories per entry
    mean_pnorms = pnorms.sum(dim=0) / cnt            # (N, P)

# ⟨ΔΔ⟩ over trajectories, NaN-aware. Since bad entries are zeroed, the einsum
# sum skips them; normalize by the per-(n) valid trajectory count.
# valid count at the (n) level for the outer-product normalization:
    ntraj_valid = valid.all(dim=-1).sum(dim=0).clamp_min(1).to(pnorms.dtype)  # (N,)

    Exx = torch.einsum('tnp,tnq->npq', pnorms, pnorms) / ntraj_valid[:, None, None]
    ExEx = torch.einsum('np,nq->npq', mean_pnorms, mean_pnorms)
    Cij_trajavg = Exx - ExEx                          # (N, P, P)

    Cij = Cij_trajavg.mean(dim=0)                     # avg over events

    if norm:
        denom = torch.sum(Cij.diag())
        chi = torch.sum(Cij) / denom if denom.abs() > 1e-12 else torch.tensor(float('nan'))
    else:
        chi = torch.sum(Cij)
    return chi



def Ldot(p1,p2,R=1,beta=1): #just the Lorentz dot product
    E1, E2 = p1.norm(dim=-1), p2.norm(dim=-1)
    dot = (p1 * p2).sum(dim=-1)
    return E1*E2 - dot

def compute_thrust_susceptibility(p0,phat,norm=True): #computes at fixed time: phats is a (Ntraj,Nsample,Nparticles,3) array
    Ntraj=phat.shape[0]
    Nsample=phat.shape[1]
    p0_sample = p0[None,:Nsample]
    axes_0 = compute_thrust_axis_torch(ThreeVecToFourVec(p0_sample))
    ndots_0 = np.einsum('tsmi,tsi->tsm',p0_sample,axes_0)
    ind_0= np.argsort(ndots_0,axis=2) # sort along particle axis

    p0_sorted = np.take_along_axis(p0_sample, ind_0[..., None], axis=2)
    phat_sorted = np.take_along_axis(phat, ind_0[..., None], axis=2)
    sigmas = torch.as_tensor(
#        np.abs(np.einsum('tsmi,tsi->tsm',p0_sorted-phat_sorted,axes_0))
        Ldot(p0_sorted,phat_sorted) #cleaner if we use a dot product comparison rather than naive momentum, but this is identical to the EMD dot susceptibility
    )

    valid = torch.isfinite(sigmas).all(dim=-1)                 # (Ntraj, Nsample): event ok if all particles finite
    sigmas = torch.where(valid[..., None], sigmas, torch.zeros_like(sigmas))

    # ⟨σσ⟩ over trajectories, NaN-aware: zeroed-out bad entries drop from the sum,
    # normalize by the count of valid trajectories per event
    cnt = valid.sum(dim=0).clamp_min(1).to(sigmas.dtype)       # (Nsample,)
    Exx  = torch.einsum('tnp,tnq->npq', sigmas, sigmas) / cnt[:, None, None]
    mean_sig = sigmas.sum(dim=0) / cnt[:, None]                # (Nsample, Nparticles)
    ExEx = torch.einsum('np,nq->npq', mean_sig, mean_sig)
    Cij_trajavg = Exx - ExEx

    Cij = Cij_trajavg.mean(dim=0)
    if norm:
        denom = torch.sum(Cij.diag())
        chi = torch.sum(Cij) / denom if denom.abs() > 1e-12 else torch.tensor(float('nan'))
    else:
        chi = torch.sum(Cij)
    return chi



def ThreeVecToFourVec(p3):
    E = torch.linalg.norm(p3,dim=-1,keepdim=True)
    p4 = torch.cat([E,p3],dim=-1)
    return np.array(p4)



def _sign_matrix(L, device, dtype):
    """All 2^(L-1) sign vectors of length L, first entry fixed to +1
    (eps and -eps give the same |sum_i eps_i p_i|, so half suffice)."""
    rows = [(1,) + tail for tail in itertools.product((1, -1), repeat=L - 1)]
    return torch.tensor(rows, device=device, dtype=dtype)

@torch.no_grad()
def compute_thrust_axis_torch(events, device='cpu', dtype=torch.float64,
                              chunk=20000, max_enum=16, n_seed=4, n_iter=20,
                              return_thrust=False):
    """Exact thrust axis (and optionally thrust) for a batch of e+e- events.
    events: (..., M, 4) = (E, px, py, pz), with any number of leading batch
    dimensions (including none).
    Returns axis (..., 3) numpy, or (axis, thrust) if return_thrust.
    """

        
    ev = torch.as_tensor(events, dtype=dtype)
    batch_shape, M = ev.shape[:-2], ev.shape[-2]
    ev = ev.reshape(-1, *ev.shape[-2:])
    B = ev.shape[0]
    ax_out = torch.empty(B, 3, dtype=dtype)
    t_out = torch.empty(B, dtype=dtype)

    def _finish(sums, denom):                       # sums: (b, K, 3)
        best = sums.norm(dim=-1).max(dim=1)         # values, indices
        axis = torch.gather(sums, 1,
                            best.indices[:, None, None].expand(-1, 1, 3)).squeeze(1)
        axis = axis / best.values.unsqueeze(-1).clamp_min(1e-30)
        # sign convention: n_z > 0, tie-broken by n_y then n_x
#        key = torch.where(axis[:, 2] != 0, axis[:, 2],
#              torch.where(axis[:, 1] != 0, axis[:, 1], axis[:, 0]))
#        axis = axis * torch.where(key < 0, -1.0, 1.0).to(dtype).unsqueeze(-1)
        return axis, best.values / denom

    if M <= max_enum:                               # exact enumeration
        S = _sign_matrix(M, device, dtype)          # (K, M), K = 2^(M-1)
        for s in range(0, B, chunk):
            p = ev[s:s+chunk, :, 1:4].to(device)
            denom = p.norm(dim=-1).sum(dim=-1)
            sums = torch.einsum('km,bmd->bkd', S, p)
            axis, T = _finish(sums, denom)
            ax_out[s:s+chunk], t_out[s:s+chunk] = axis.cpu(), T.cpu()
    else:                                           # seeded fixed-point fallback
        S = _sign_matrix(n_seed, device, dtype)
        for s in range(0, B, chunk):
            p = ev[s:s+chunk, :, 1:4].to(device)
            norms = p.norm(dim=-1)
            denom = norms.sum(dim=-1)
            lead = norms.argsort(dim=-1, descending=True)[:, :n_seed]
            p_lead = torch.gather(p, 1, lead.unsqueeze(-1).expand(-1, -1, 3))
            axes = torch.einsum('km,bmd->bkd', S, p_lead)
            axes = axes / axes.norm(dim=-1, keepdim=True).clamp_min(1e-30)
            for _ in range(n_iter):
                eps = torch.sign(torch.einsum('bmd,bkd->bkm', p, axes))
                sums = torch.einsum('bkm,bmd->bkd', eps, p)
                new_axes = sums / sums.norm(dim=-1, keepdim=True).clamp_min(1e-30)
                if torch.allclose(new_axes, axes):
                    axes = new_axes
                    break
                axes = new_axes
            axis, T = _finish(sums, denom)
            ax_out[s:s+chunk], t_out[s:s+chunk] = axis.cpu(), T.cpu()

    ax_out = ax_out.reshape(*batch_shape, 3)
    if return_thrust:
        return ax_out.numpy(), t_out.reshape(batch_shape).numpy()
    return ax_out.numpy()

def thrust_similarity(p0, phat):
    """p0: (Nsample, P, 3) clean;  phat: (Ntraj, Nsample, P, 3) denoised.
       Returns |cos(angle)| between clean and reconstructed thrust axes,
       averaged over trajectories and events (1 = perfectly aligned)."""
    p0   = p0.double()                  # or .float() — just pick one and use it for both
    phat = phat.double()
    n0   = compute_thrust_axis_torch(ThreeVecToFourVec(p0[:phat.shape[1]]))       # (Nsample, 3)
    nhat = compute_thrust_axis_torch(ThreeVecToFourVec(phat))                      # (Ntraj, Nsample, 3)
    
    cos2 = np.einsum('si,tsi->ts',n0,nhat)**2

   # cos = torch.einsum('tnd,nd->tn', nhat, n0).abs().clamp(max=1.0)  # (Ntraj, Nsample)

    valid = np.isfinite(cos2)                             # numpy
    cos2 = np.where(valid, cos2, 0.0)                     # zero out bad entries
    return cos2.sum() / np.clip(valid.sum(), 1, None)


#n0 = torch.from_numpy(compute_thrust_axis_torch(ThreeVecToFourVec(pspace_dat[:10000].double())))
#print((torch.einsum('nd,nd->n', n0, n0).abs()).mean())   # must be 1.0

sims_5=[]
for i in range(len(ts_forward_5)):
    sims_5.append(thrust_similarity(pspace_dat_5,phats_5[i]))

print(sims_5)

sims_10=[]
for i in range(len(ts_forward_10)):
    sims_10.append(thrust_similarity(pspace_dat_10,phats_10[i]))

print(sims_10)

sims_20=[]
for i in range(len(ts_forward_20)):
    sims_20.append(thrust_similarity(pspace_dat_20,phats_20[i]))

print(sims_20)

sims_25=[]
for i in range(len(ts_forward_25)):
    sims_25.append(thrust_similarity(pspace_dat_25,phats_25[i]))

print(sims_25)

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(ts_forward_5,  sims_5,  'o-', linewidth=2, markersize=6, label='N=20,xi=0.5')
ax.plot(ts_forward_10, sims_10, 's-', linewidth=2, markersize=6, label='N=20,xi=5')
ax.plot(ts_forward_20, sims_20, '^-', linewidth=2, markersize=6, label='N=20,xi=10')
ax.plot(ts_forward_25, sims_25, '^-', linewidth=2, markersize=6, label='N=20,xi=1.4')

ax.set_xlabel('t*', fontsize=13)
#ax.set_ylabel(r'$\chi(t^*)$', fontsize=13)
ax.set_title('Thurst Similarity across multiplicity', fontsize=13)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
plt.tight_layout()
#plt.savefig('/Users/shashin/Downloads/SARGE/thurst_similarity_new1.png', dpi=150)
plt.show()

chis_norm_5=[]

for i in range(len(ts_forward_5)):
    chi = compute_thrust_susceptibility(pspace_dat_5,phats_5[i])
    chis_norm_5.append(chi)
   # chi = compute_thrust_susceptibility(pspace_dat,phats[i],norm=False)
   # chis_unnorm.append(chi)

print(chis_norm_5)

chis_norm_10=[]

for i in range(len(ts_forward_10)):
    chi = compute_thrust_susceptibility(pspace_dat_10,phats_10[i])
    chis_norm_10.append(chi)
   # chi = compute_thrust_susceptibility(pspace_dat,phats[i],norm=False)
   # chis_unnorm.append(chi)

print(chis_norm_10)

chis_norm_20=[]

for i in range(len(ts_forward_20)):
    chi = compute_thrust_susceptibility(pspace_dat_20,phats_20[i])
    chis_norm_20.append(chi)
   # chi = compute_thrust_susceptibility(pspace_dat,phats[i],norm=False)
   # chis_unnorm.append(chi)

print(chis_norm_20)

chis_norm_25=[]

for i in range(len(ts_forward_25)):
    chi = compute_thrust_susceptibility(pspace_dat_25,phats_25[i])
    chis_norm_25.append(chi)
   # chi = compute_thrust_susceptibility(pspace_dat,phats[i],norm=False)
   # chis_unnorm.append(chi)

print(chis_norm_25)

# %%
#plt.plot(ts_forward,chis_norm,label='normalized')
#plt.plot(ts_forward,30*np.array(chis_unnorm),label='un-normalized (rescaled)')
#plt.title('momentum susceptibility, tgaus=0, T=5000')
#plt.xlabel('t')
#plt.ylabel(r'$\chi(t)$')
#plt.legend(loc='upper right')

fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(ts_forward_5,  chis_norm_5,  'o-', linewidth=2, markersize=6, label='N=15,xi=1.4')
ax.plot(ts_forward_10, chis_norm_10, 's-', linewidth=2, markersize=6, label='N=18,xi=1.4')
ax.plot(ts_forward_20, chis_norm_20, '^-', linewidth=2, markersize=6, label='N=24,xi=1.4')
#ax.plot(ts_forward_25, chis_norm_25, '^-', linewidth=2, markersize=6, label='N=20,xi=1.4')

ax.set_xlabel('t*', fontsize=13)
ax.set_ylabel(r'$\chi(t^*)$', fontsize=13)
ax.set_title('Susceptibility ', fontsize=13)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
plt.tight_layout()
#plt.savefig('/Users/shashin/Downloads/SARGE/thrust_susceptibility_new_1.png', dpi=150)
plt.show()
# %%
```

```python
from scipy.interpolate import PchipInterpolator
import numpy as np
import torch
import matplotlib.pyplot as plt


# -------------------------------------------------------
# Bootstrap uncertainty on susceptibility
# -------------------------------------------------------

def bootstrap_thrust_susceptibility(
    p0,
    phat,
    n_boot=200,
    seed=1234,
    resample_trajectories=True
):
    """
    Estimate susceptibility and its uncertainty by bootstrap resampling.

    p0:
        (Nsample, Nparticles, 3)

    phat:
        (Ntraj, Nsample, Nparticles, 3)

    Returns
    -------
    chi : float
        Susceptibility evaluated on the full dataset.

    chi_err : float
        Bootstrap standard deviation.
    """

    # Central value from full sample
    chi = float(compute_thrust_susceptibility(p0, phat))

    rng = np.random.default_rng(seed)

    Ntraj = phat.shape[0]
    Nsample = phat.shape[1]

    boot_values = []

    for _ in range(n_boot):

        # Resample events
        sample_idx = torch.from_numpy(
            rng.integers(0, Nsample, size=Nsample)
        ).long()

        p0_boot = p0[sample_idx]
        phat_boot = phat[:, sample_idx]

        # Optionally also bootstrap stochastic reverse trajectories
        if resample_trajectories:
            traj_idx = torch.from_numpy(
                rng.integers(0, Ntraj, size=Ntraj)
            ).long()

            phat_boot = phat_boot[traj_idx]

        chi_boot = compute_thrust_susceptibility(
            p0_boot,
            phat_boot
        )

        if torch.is_tensor(chi_boot):
            chi_boot = chi_boot.item()

        if np.isfinite(chi_boot):
            boot_values.append(chi_boot)

    chi_err = np.std(boot_values, ddof=1)

    return chi, chi_err


# -------------------------------------------------------
# Calculate susceptibility + errors
# -------------------------------------------------------

def susceptibility_curve(p0, phats, n_boot=5):

    chis = []
    errs = []

    for phat in phats:

        chi, err = bootstrap_thrust_susceptibility(
            p0,
            phat,
            n_boot=n_boot
        )

        chis.append(chi)
        errs.append(err)

    return np.array(chis), np.array(errs)


chis_norm_5, errs_5 = susceptibility_curve(
    pspace_dat_5,
    phats_5
)

chis_norm_10, errs_10 = susceptibility_curve(
    pspace_dat_10,
    phats_10
)

chis_norm_20, errs_20 = susceptibility_curve(
    pspace_dat_20,
    phats_20
)

#chis_norm_25, errs_25 = susceptibility_curve(
#    pspace_dat_25,
#    phats_25
#)


# -------------------------------------------------------
# Smooth PCHIP interpolation
# -------------------------------------------------------

def smooth_curve(x, y, npoints=10):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    x_smooth = np.linspace(x.min(), x.max(), npoints)

    interp = PchipInterpolator(x, y)
    y_smooth = interp(x_smooth)

    return x_smooth, y_smooth


# -------------------------------------------------------
# Plot
# -------------------------------------------------------

fig, ax = plt.subplots(figsize=(8, 5))

datasets = [
    (ts_forward_5,  chis_norm_5,  errs_5,  'o', r'$N=5,\ \xi=88$'),
    (ts_forward_10, chis_norm_10, errs_10, 's', r'$N=10,\ \xi=11$'),
    (ts_forward_25, chis_norm_25, errs_25, '^', r'$N=20,\ \xi=1.4$'),
   # (ts_forward_20, chis_norm_20, errs_20, 'D', r'$N=20,\ \xi=1.4$'),
]

for t, chi, err, marker, label in datasets:

    t = np.asarray(t, dtype=float)
    chi = np.asarray(chi, dtype=float)
    err = np.asarray(err, dtype=float)

    # Smooth interpolating curve
    #t_smooth, chi_smooth = smooth_curve(t, chi)

    line, = ax.plot(
        t,
        chi,
        linewidth=2,
        label=label
    )

    # Actual measured points + error bars
    ax.errorbar(
        t,
        chi,
        yerr=err,
        fmt=marker,
        markersize=6,
        capsize=3,
        linestyle='none',
        color=line.get_color()
    )


# chi = 1 reference
ax.axhline(
    1.0,
    linestyle='--',
    linewidth=1.2,
    alpha=0.6
)

ax.set_xlabel(r'$t^*$', fontsize=14)
ax.set_ylabel(r'$\chi(t^*)$', fontsize=14)

ax.set_title(
    'Susceptibility across diffusion scale',
    fontsize=14
)

ax.grid(True, alpha=0.25)
ax.legend(fontsize=11)

ax.tick_params(
    axis='both',
    labelsize=11
)

plt.tight_layout()

plt.savefig(
    '/Users/shashin/Downloads/SARGE/thrust_susceptibility_N5_N10_N20.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
import numpy as np
import torch
import matplotlib.pyplot as plt


# -------------------------------------------------------
# Bootstrap uncertainty on thrust similarity
# -------------------------------------------------------

def bootstrap_thrust_similarity(
    p0,
    phat,
    n_boot=200,
    seed=1234,
    resample_trajectories=True
):
    """
    Estimate thrust similarity and its uncertainty by bootstrap resampling.

    p0:
        (Nsample, Nparticles, 3)

    phat:
        (Ntraj, Nsample, Nparticles, 3)

    Returns
    -------
    sim : float
        Thrust similarity evaluated on the full dataset.

    sim_err : float
        Bootstrap standard deviation.
    """

    # Central value from full sample
    sim = float(thrust_similarity(p0, phat))

    rng = np.random.default_rng(seed)

    Ntraj = phat.shape[0]
    Nsample = phat.shape[1]

    boot_values = []

    for _ in range(n_boot):

        # Resample events
        sample_idx = torch.from_numpy(
            rng.integers(0, Nsample, size=Nsample)
        ).long()

        p0_boot = p0[sample_idx]
        phat_boot = phat[:, sample_idx]

        # Optionally also resample reverse trajectories
        if resample_trajectories:
            traj_idx = torch.from_numpy(
                rng.integers(0, Ntraj, size=Ntraj)
            ).long()

            phat_boot = phat_boot[traj_idx]

        sim_boot = thrust_similarity(
            p0_boot,
            phat_boot
        )

        if np.isfinite(sim_boot):
            boot_values.append(sim_boot)

    sim_err = np.std(boot_values, ddof=1)

    return sim, sim_err


# -------------------------------------------------------
# Calculate similarity + errors
# -------------------------------------------------------

def thrust_similarity_curve(p0, phats, n_boot=5):

    sims = []
    errs = []

    for phat in phats:

        sim, err = bootstrap_thrust_similarity(
            p0,
            phat,
            n_boot=n_boot
        )

        sims.append(sim)
        errs.append(err)

    return np.array(sims), np.array(errs)


sims_5, errs_sim_5 = thrust_similarity_curve(
    pspace_dat_5,
    phats_5
)

sims_10, errs_sim_10 = thrust_similarity_curve(
    pspace_dat_10,
    phats_10
)

sims_20, errs_sim_20 = thrust_similarity_curve(
    pspace_dat_20,
    phats_20
)

sims_25, errs_sim_25 = thrust_similarity_curve(
    pspace_dat_25,
    phats_25
)


# -------------------------------------------------------
# Plot
# -------------------------------------------------------

fig, ax = plt.subplots(figsize=(8, 5))

datasets = [
    (ts_forward_5,  sims_5,  errs_sim_5,  'o', r'$N=5,\ \xi=88$'),
    (ts_forward_10, sims_10, errs_sim_10, 's', r'$N=10,\ \xi=11$'),
    (ts_forward_25, sims_25, errs_sim_25, '^', r'$N=20,\ \xi=1.4$'),
   # (ts_forward_20, sims_20, errs_sim_20, 'D', r'$N=24,\ \xi=1.4$'),
]

for t, sim, err, marker, label in datasets:

    ax.errorbar(
        t,
        sim,
        yerr=err,
        fmt=marker + '-',
        linewidth=2,
        markersize=6,
        capsize=3,
        label=label
    )

ax.set_xlabel(r'$t^*$', fontsize=14)
ax.set_ylabel(r'$\langle (\hat{n}_0 \cdot \hat{n}_{\rm rec})^2 \rangle$', fontsize=14)

ax.set_title(
    'Thrust similarity across diffusion scale',
    fontsize=14
)

ax.grid(True, alpha=0.25)
ax.legend(fontsize=11)

ax.tick_params(
    axis='both',
    labelsize=11
)

plt.tight_layout()

plt.savefig(
    '/Users/shashin/Downloads/SARGE/thrust_similarity_newwst1.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
